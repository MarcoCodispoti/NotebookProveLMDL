Implementazione Permutation importance

```python
from sklearn.inspection import permutation_importance

result = permutation_importance(
    model, X_val, y_val,
    n_repeats=10, random_state=42, scoring="accuracy"
)

perm_importances_mean = result.importances_mean
perm_importances_std = result.importances_std
```

Implementazione PDP

```python
from sklearn.inspection import PartialDependenceDisplay

PartialDependenceDisplay.from_estimator(
    model, X_train, features=["eta"],
    kind="both", # "average" per il solo PDP, "both" per PDP + ICE
    grid_resolution=100
)
```
